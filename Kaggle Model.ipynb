```python
import os
import cv2
import re
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from torchvision import models
from sklearn.model_selection import GroupShuffleSplit
import matplotlib.pyplot as plt
import time
from tqdm import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Training on: {device}")

IMG_SIZE = (224, 224)
BATCH_SIZE = 256  
NUM_EPOCHS = 5
-
def find_dataset_root():
    possible_roots = [
        "/kaggle/input/columbian-dataset/columbia_gaze",
        "/kaggle/input/columbia-gaze-dataset",
        "/kaggle/input"
    ]
    for root in possible_roots:
        if os.path.exists(root):
            for root_dir, dirs, files in os.walk(root):
                if "0001" in dirs: return root_dir
    return None

DATA_DIR = find_dataset_root()
if not DATA_DIR: raise ValueError("Dataset not found!")

def define_engagement_class(gaze_h, gaze_v, head_h, head_v):
    attentive_thresh = 15
    distracted_thresh = 30
    if abs(gaze_v) > 10: return 2 # Disengaged
    if abs(gaze_h) > distracted_thresh or abs(head_h) > distracted_thresh: return 1 # Distracted
    if abs(gaze_h) <= attentive_thresh and abs(head_h) <= attentive_thresh: return 0 # Attentive
    return 1 

def load_data_from_filenames(root_dir):
    data = []
    pattern = re.compile(r'^(\d+)_(\w+)_(-?\d+)P_(-?\d+)V_(-?\d+)H\.jpg$')
    all_files = []
    for dirpath, _, filenames in os.walk(root_dir):
        for f in filenames:
            if f.endswith('.jpg'): all_files.append((dirpath, f))
            
    for dirpath, filename in tqdm(all_files, desc="Parsing Labels"):
        match = pattern.match(filename)
        if match:
            data.append({
                'subject_id': int(match.group(1)),
                'full_path': os.path.join(dirpath, filename),
                'engagement_class': define_engagement_class(
                    int(match.group(5)), int(match.group(4)), int(match.group(3)), 0
                )
            })
    return pd.DataFrame(data)

# ---  CACHED DATASET (THE FIX) ---
class CachedEyeGazeDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe
        self.transform = transform
        self.images = []
        self.labels = []
        
        print(f"Loading and resizing {len(dataframe)} images to RAM... (This takes ~1 min)")
   
        for i in tqdm(range(len(dataframe))):
            row = dataframe.iloc[i]
            try:
                img = cv2.imread(row['full_path'])
                if img is not None:
                    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    img = cv2.resize(img, IMG_SIZE) 
                    self.images.append(img)
                    self.labels.append(row['engagement_class'])
            except Exception as e:
                pass

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        image = self.images[idx]
        label = self.labels[idx]
        
        if self.transform:
            image = self.transform(image)
            
        return image, label

# Load Data
full_df = load_data_from_filenames(DATA_DIR)
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(full_df, groups=full_df['subject_id']))
train_df = full_df.iloc[train_idx].reset_index(drop=True)
test_df = full_df.iloc[test_idx].reset_index(drop=True)

# Transforms (No Resize needed here, done in init)
train_transforms = transforms.Compose([
    transforms.ToPILImage(),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

test_transforms = transforms.Compose([
    transforms.ToPILImage(),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

print("Processing Training Set...")
train_dataset = CachedEyeGazeDataset(train_df, transform=train_transforms)
print("Processing Test Set...")
test_dataset = CachedEyeGazeDataset(test_df, transform=test_transforms)

# DataLoaders (num_workers=0 is safer for cached RAM data)
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
dataloaders = {'train': train_loader, 'test': test_loader}

def get_model(model_name):
    if model_name == 'resnet18':
        model = models.resnet18(weights='IMAGENET1K_V1')
        model.fc = nn.Linear(model.fc.in_features, 3)
    elif model_name == 'mobilenet_v2':
        model = models.mobilenet_v2(weights='IMAGENET1K_V1')
        model.classifier[1] = nn.Linear(model.last_channel, 3)
    elif model_name == 'resnet50':
        model = models.resnet50(weights='IMAGENET1K_V1')
        model.fc = nn.Linear(model.fc.in_features, 3)
        
    model = model.to(device)
    if torch.cuda.device_count() > 1:
        model = nn.DataParallel(model)
    return model

def train_model(model, criterion, optimizer, num_epochs=5):
    history = []
    for epoch in range(num_epochs):
        print(f'\nEpoch {epoch+1}/{num_epochs}')
        
        for phase in ['train', 'test']:
            if phase == 'train': model.train()
            else: model.eval()

            running_loss = 0.0
            running_corrects = 0
            
            pbar = tqdm(dataloaders[phase], desc=f"{phase}", unit="batch")
            
            for inputs, labels in pbar:
                inputs = inputs.to(device)
                labels = labels.to(device)

                optimizer.zero_grad()
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
                pbar.set_postfix(loss=loss.item())

            epoch_acc = running_corrects.double() / len(dataloaders[phase].dataset)
            print(f'{phase} Acc: {epoch_acc:.4f}')
            if phase == 'test': history.append(epoch_acc.cpu().item())
            
    return model, history

results = {}
models_to_test = ['resnet18', 'mobilenet_v2', 'resnet50']

for name in models_to_test:
    print(f"\n--- Training {name} ---")
    model = get_model(name)
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    trained_model, hist = train_model(model, nn.CrossEntropyLoss(), optimizer, num_epochs=NUM_EPOCHS)
    results[name] = hist
    torch.save(trained_model.state_dict(), f"{name}.pth")

# Final Plot
plt.figure(figsize=(10, 6))
for name, hist in results.items():
    plt.plot(hist, label=name)
plt.title("Model Comparison (Cached Data)")
plt.legend()
plt.savefig("comparison.png")
plt.show()
```
